```python
cleaned_df = df.copy()
```

```python
cleaned_df.head()
```

```python
cleaned_df["Order_Date"] = pd.to_datetime(
    cleaned_df["Order_Date"],
    errors="coerce"
)
```

```python
cleaned_df["Order_Date"].dtype
```

```python
cleaned_df["Age"] = cleaned_df["Age"].fillna(
    cleaned_df["Age"].median()
)
```

```python
cleaned_df["Age"].isnull().sum()
```

```python
cleaned_df["City"] = cleaned_df["City"].fillna("Unknown")
```

```python
text_columns = [
    "Order_ID",
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "City",
    "Product",
    "Category"
]

for col in text_columns:
    cleaned_df[col] = cleaned_df[col].astype(str).str.strip()
```

```python
print("Invalid Age:")
print(cleaned_df[(cleaned_df["Age"] < 18) | (cleaned_df["Age"] > 65)])

print("\nInvalid Quantity:")
print(cleaned_df[cleaned_df["Quantity"] <= 0])

print("\nInvalid Unit Price:")
print(cleaned_df[cleaned_df["Unit_Price"] <= 0])

print("\nInvalid Total Sales:")
print(cleaned_df[cleaned_df["Total_Sales"] <= 0])
```

```python
cleaned_df["Calculated_Sales"] = (
    cleaned_df["Quantity"] * cleaned_df["Unit_Price"]
)

cleaned_df["Sales_Difference"] = (
    cleaned_df["Total_Sales"] - cleaned_df["Calculated_Sales"]
).abs()
```

```python
cleaned_df["Sales_Difference"].describe()
```

```python
print("Gender values:")
print(cleaned_df["Gender"].value_counts())

print("\nCity values:")
print(cleaned_df["City"].value_counts())

print("\nCategory values:")
print(cleaned_df["Category"].value_counts())

print("\nProduct values:")
print(cleaned_df["Product"].value_counts())
```

```python
numeric_columns = [
    "Age",
    "Quantity",
    "Unit_Price",
    "Total_Sales"
]

for col in numeric_columns:
    Q1 = cleaned_df[col].quantile(0.25)
    Q3 = cleaned_df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = cleaned_df[
        (cleaned_df[col] < lower) |
        (cleaned_df[col] > upper)
    ]

    print(f"{col}: {len(outliers)} outliers")
```

```python
cleaned_df = cleaned_df.drop(
    columns=["Calculated_Sales", "Sales_Difference"],
    errors="ignore"
)
```

```python
print("Shape:", cleaned_df.shape)

print("\nMissing values:")
print(cleaned_df.isnull().sum())

print("\nDuplicate rows:")
print(cleaned_df.duplicated().sum())

print("\nData types:")
print(cleaned_df.dtypes)
```

```python
output_file = "/content/ApexPlanet_Task1_Cleaned_Dataset.xlsx"

cleaned_df.to_excel(
    output_file,
    index=False
)

print("Cleaned dataset saved successfully!")
```
